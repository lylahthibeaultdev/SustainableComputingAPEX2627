**Purpose:**
This program implements algorithms intentionally designed several ways with different aspects in mind, each prioritizing either traditional algorithmic efficiency/time complexity or developed in a way that may be less efficient, but impelements aspects of "green" computing, as outlined by the green algorithms initiative, and with technical implementation supplied from other sources.

By testing the runtime, actual computer (hardware) temperature, and estimated carbon emissions supplied by the code-carbon tool, we will be supplied with an interesting data set comparing the actual output and differences in efficiency for each algorithm with these different aspects prioritized. These algorithms have been chosen as they are able to be implemented simply enough to keep data and findings straightforward, while also having applications to LLM research because these search aglorithms (DFS, BFS) are utilized, or are the basis of search algorithms used in AI in order to train and adapt with the central idea of "searching" for possible paths for a problem solution in an efficient manner.

Currently, the priority has been placed on LLM algorithms to be as efficient as possible, in order to create the most powerful and useful models. However with the rapid development and adoption of AI, we've quickly began to see the environmental implications of suppporting these models. The aim of this project is to study the handoffs of either prioritizing technical efficency or energy efficiency, and how those two aspects interact with each other, as well as investigating energy efficiency on the scale of hardware (computer temp) vs the larger scale of estimated carbon emissions. 

The scale of this project does not reach the scope of hardware utilized, or algorithms run to uphold LLM architecture, but is aimed to study the aspects of green computing as it emerges and use this data to envision the large scale application of these findings. 

This experiment may be repeated using the UMass CIS Cluster Machine in order to show with scale to a larger dataset, and different hardware.
